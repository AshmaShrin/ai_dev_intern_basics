## Day 1: Classes and objects

**Big idea:** a **class** is a blueprint. An **object** (instance) is a thing built from that blueprint.
- Blueprint of a car -> `Car` class
- Your red car, my blue car -> two **objects**

An object has:
- **attributes**: data it remembers (colour, speed)
- **methods**: things it can do (drive, stop). These are functions that live inside a class.

`__init__` runs automatically when you create an object. `self` means "this very object".

````python
class Dog:
    def __init__(self, name, age):
        self.name = name      # attribute
        self.age = age

    def bark(self):           # method
        return f"{self.name} says Woof!"

    def birthday(self):
        self.age += 1

rex = Dog("Rex", 3)       # create an object
luna = Dog("Luna", 5)

print(rex.name, rex.age)
print(rex.bark())
rex.birthday()
print(rex.age)            # 4
print(luna.age)           # still 5. Objects are independent
````

## Day 2: Nice printing, class attributes, encapsulation

### `__str__` and `__repr__`
By default, printing an object looks ugly (`<__main__.Dog object at 0x...>`). Special methods fix this:
- `__str__` -> friendly text for users (`print(obj)`)
- `__repr__` -> clear text for developers (what you see when you just type `obj`)

````python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"

    def __repr__(self):
        return f"Book(title={self.title!r}, author={self.author!r})"

b = Book("Python Basics", "Guido")
print(b)
b
````

### Class attributes vs instance attributes
- **Instance attribute**: belongs to one object (`self.name`).
- **Class attribute**: shared by all objects, defined in the class body.

````python
class Student:
    school = "AI Academy"          # class attribute (shared)
    count = 0

    def __init__(self, name):
        self.name = name           # instance attribute
        Student.count += 1

a = Student("Asha")
b = Student("Ravi")
print(a.school, b.school, Student.count)
````

### Encapsulation and `@property`
Encapsulation = keep data safe inside the object and control how it is changed.
Python convention: a name starting with `_` means "private, please don't touch directly".

`@property` lets a method be used like an attribute, and lets you add validation.

````python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance      # 'private' by convention

    @property
    def balance(self):               # read it like account.balance
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

acc = Account("Asha", 100)
acc.deposit(50)
print(acc.balance)
# acc.balance = 1_000_000   # error! no setter, so nobody can cheat
````

## Day 3: Inheritance

A **child class** can reuse everything from a **parent class** and add or change things. This avoids copy-pasting code.
`super()` calls the parent's version of a method.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

    def intro(self):
        return f"I am {self.name} and I say {self.speak()}"

class Cat(Animal):             # Cat inherits from Animal
    def speak(self):           # override the parent's method
        return "Meow"

class Dog(Animal):
    def __init__(self, name, tricks):
        super().__init__(name) # reuse parent's __init__
        self.tricks = tricks

    def speak(self):
        return "Woof"

print(Cat("Tom").intro())
d = Dog("Rex", ["sit", "roll"])
print(d.intro(), d.tricks)
print(isinstance(d, Animal), isinstance(d, Cat))
````

## Day 4: Polymorphism, composition, dunder methods

### Polymorphism
"Many forms": different classes offer the **same method name**, so code can treat them the same way, without caring which exact class it is.

````python
class Circle:
    def __init__(self, r): self.r = r
    def area(self): return 3.14159 * self.r ** 2

class Square:
    def __init__(self, s): self.s = s
    def area(self): return self.s ** 2

shapes = [Circle(1), Square(2), Circle(2)]
for shape in shapes:
    print(type(shape).__name__, shape.area())   # same call, different behaviour
````

### Composition: "has-a"
Inheritance means "is-a" (a Dog **is an** Animal). Composition means "has-a": an object **contains** other objects (a Car **has an** Engine). Often simpler and more flexible than inheritance.

````python
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine()    # Car HAS an Engine

    def start(self):
        return "Car: " + self.engine.start()

print(Car().start())
````

### Dunder (magic) methods
Names with double underscores let your objects work with Python's operators:

| Method | Enables |
|---|---|
| `__add__` | `a + b` |
| `__eq__` | `a == b` |
| `__len__` | `len(a)` |
| `__getitem__` | `a[0]` |

````python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

Vector(1, 2) + Vector(3, 4)
````

## Day 5: `@classmethod`, `@staticmethod`, dataclasses

- **`@classmethod`** gets the class (`cls`) instead of an object. Often used as an alternative constructor.
- **`@staticmethod`** is just a normal function placed inside the class for organisation. No `self`, no `cls`.

````python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @classmethod
    def from_fahrenheit(cls, f):
        return cls((f - 32) * 5 / 9)

    @staticmethod
    def is_freezing(celsius):
        return celsius <= 0

t = Temperature.from_fahrenheit(212)
print(t.celsius)                      # 100.0
print(Temperature.is_freezing(-3))
````

### Dataclasses
When a class mostly stores data, `@dataclass` writes `__init__`, `__repr__` and `__eq__` for you.

````python
from dataclasses import dataclass

@dataclass
class Point:
    x: float
    y: float = 0.0

p = Point(3, 4)
print(p, p == Point(3, 4))
````

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_borrowed = False

class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed = []

    def borrowed_titles(self):
        return [b.title for b in self.borrowed]

class Library:
    def __init__(self):
        self.books = []
        self.members = {}

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members[member.name] = member

    def _find(self, title):
        for b in self.books:
            if b.title == title:
                return b
        raise ValueError('No such book')

    def borrow(self, member_name, title):
        book = self._find(title)
        if book.is_borrowed:
            raise ValueError('Already borrowed')
        book.is_borrowed = True
        self.members[member_name].borrowed.append(book)

    def give_back(self, member_name, title):
        book = self._find(title)
        book.is_borrowed = False
        self.members[member_name].borrowed.remove(book)

    def available_titles(self):
        return [b.title for b in self.books if not b.is_borrowed]
```
