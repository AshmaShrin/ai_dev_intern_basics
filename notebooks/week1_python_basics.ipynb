## Day 3: Loops, lists and tuples

### Lists
A **list** holds many values in order. It can change (**mutable**).

````python
fruits = ["apple", "banana", "mango"]
fruits.append("orange")      # add at the end
fruits.remove("banana")      # remove by value
print(fruits)
print(fruits[0], fruits[-1], len(fruits))
print(fruits[1:3])           # slicing works like strings
print("mango" in fruits)     # True
````

### `for` loops
Repeat something for each item. `range(5)` gives `0,1,2,3,4`.

````python
for fruit in fruits:
    print("I like", fruit)

for i in range(5):
    print("Round", i)

total = 0
for number in [1, 2, 3, 4]:
    total += number          # same as total = total + number
print("Total:", total)
````

### `while` loops
Repeat **as long as** a condition is true. Be careful: make sure it can stop, or it runs forever!

````python
count = 3
while count > 0:
    print("Countdown", count)
    count -= 1
print("Liftoff!")
````

### Tuples
A **tuple** is like a list that cannot change (**immutable**). Written with `()`. Good for fixed data like coordinates.

````python
point = (3, 4)
x, y = point            # 'unpacking'
print(x, y)
# point[0] = 10         # would give an error. Tuples can't change!
````

## Day 4: Dictionaries, sets, comprehensions, functions

### Dictionaries
A **dict** stores `key: value` pairs. Like a real dictionary: look up a word (key) to get its meaning (value).

````python
student = {"name": "Ravi", "age": 20, "marks": [80, 90]}
print(student["name"])
student["age"] = 21                 # change
student["city"] = "Pune"            # add
print(student.get("phone", "no phone saved"))   # safe lookup with default

for key, value in student.items():
    print(key, "->", value)
````

### Sets
A **set** keeps only unique values, no order. Great for removing duplicates.

````python
tags = {"ai", "python", "ai", "data"}
print(tags)                  # duplicates disappear
print(set([1, 1, 2, 3, 3]))
````

### List comprehensions
A short way to build a list from another one.
`[expression for item in iterable if condition]`

````python
squares = [n * n for n in range(1, 6)]
evens = [n for n in range(10) if n % 2 == 0]
print(squares, evens)
````

### Functions
A **function** is a reusable block of code. `def` creates it, `return` sends back a result.

````python
def greet(name, greeting="Hello"):
    """Return a greeting message."""
    return f"{greeting}, {name}!"

print(greet("Asha"))
print(greet("Asha", greeting="Vanakkam"))
print(greet(greeting="Hi", name="Ravi"))   # keyword arguments
````

### Errors and `try / except`
Errors (exceptions) are normal. **Read the message**: the last line tells you what went wrong, and the line number tells you where. You can catch errors so your program does not crash.

````python
try:
    number = int("hello")
except ValueError:
    print("That was not a number!")

try:
    print(10 / 0)
except ZeroDivisionError as err:
    print("Oops:", err)
````

def add_mark(students, name, mark):
    if name not in students:
        students[name] = []
    students[name].append(mark)

def student_average(students, name):
    marks = students.get(name, [])
    if not marks:
        return 0
    return sum(marks) / len(marks)

def top_student(students):
    best_name = None
    best_avg = -1
    for name in students:
        avg = student_average(students, name)
        if avg > best_avg:
            best_name, best_avg = name, avg
    return best_name

def report(students):
    lines = []
    for name in students:
        avg = student_average(students, name)
        grade = 'A' if avg >= 90 else 'B' if avg >= 70 else 'C'
        lines.append(f"{name}: {avg} ({grade})")
    return lines
